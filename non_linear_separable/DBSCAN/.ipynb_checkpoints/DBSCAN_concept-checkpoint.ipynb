```python
DBSCAN  stands for '''DENSITY BASED SPETIAL CLUSTERING of APPLICATION with NOISE'''
--> jha density ziada hoti us ko ek cluster m daal deti h

              
----> it is good for non -linear data and can easily detect outliers
```

```python
is ka ader kuch cheezy ati hn such as:
----<> Min point
----<> abselon (which is radius)
----<> core point
----<> boundary point
----<> noise point
```

```python
hm ek data point pick krty hn and us k ird gird circle bnaty hn jis ka radius abselon (jo k hm bnatyn
gy) k equal ho ga and us k ander jo total points ayn gy vo utny hon gy jo hm n min points btay hin 
gy, e.g: Abselon is 0.5 and min points are 4 and data point pich kia say A, to A k ird gird 0.4 
radius ka circle bnay gy or us circle k ander A samet minimum 4 points a skty hn, ab core point s
murad hmara center vala data point h jis ko beech m rkh kr hm n abselon k size k radius vala circle
bnaya h or is circle k andr minimum points ki requirement bhi puri ho rhi ho, lekin ek point 
tha 'A' us k ird gird epsilon k size ka radius vala circle bnaya and us k andr minimum utny points 
th jitny hm n set kiye huy th to 'A' ek core point ho ga, agr A k circle k andr Minimuym points s 
km points hn to 'A' core point nhi ho ga, ya to vo boundary point ho ga ya Noise ho ga

ab y kesy dekhnah k vo vala point noise point h ya boundary point h , to is k liye vo data point 
jo k kisi core point ka neighbor hua to vo boundary point ho ga (yani vo data point kisi core point
k circle m a rha ho ga)
and noise point vo points hon gy jo na to core point ho ga na boundary point and this is the best
advantage of DBSCAN because it is robust to outliers. 
```
